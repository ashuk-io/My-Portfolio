```python
def vacuum_cleaner():
  loc = input("Enter the location: (A/B) ").upper()
  a = input("Room A(Clean/Dirty)").capitalize()
  b = input("Room B(Clean/Dirty)").capitalize()

  if loc == 'A':
    if a == 'Dirty':
      print("Clean A")
      a = 'Clean'
    if b == 'Dirty':
      print("Move to B and clean B")

      b = 'Clean'
  else:
    if b == 'Dirty':
        print("Clean B")
        b = 'Clean'
    if a == 'Dirty':
        print("Move to A and clean A")
        a = 'Clean'
  print("Final state A:- ",a,"B:-",b)

vacuum_cleaner()
```
